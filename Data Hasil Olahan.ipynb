## **4. EKSPLORASI DATA**

```python
# .head(5) = tampilkan 5 baris pertama
df.head(5)
```

```python
# Lihat nama semua kolom
print('=== Daftar Kolom ===')
for i, kolom in enumerate(df.columns):
    print(f'{i+1}. {kolom}')
```

```python
# Nama kolom aslinya sangat panjang, ganti jadi lebih pendek
# simpan di kolom baru supaya data aslinya aman

df['Q1_Memahami']      = df['AI membantu saya memahami materi kuliah yang sulit dengan lebih cepat. ']
df['Q2_Tugas']         = df['Penggunaan AI meningkatkan kualitas tugas dan laporan saya. ']
df['Q3_Ujian']         = df['Saya lebih percaya diri dalam ujian setelah menggunakan AI untuk latihan. ']
df['Q4_BelajarMandiri']= df['AI mendukung pembelajaran mandiri di luar jam kuliah ']
df['Q5_Privasi']       = df['Penggunaan AI mempengaruhi Privasi Data saya']
df['Q6_BerpikirKritis']= df['Saya khawatir AI mengurangi kemampuan berpikir kritis saya. ']
df['Q7_Kampus']        = df['Kampus ITSB sudah siap mendukung penggunaan AI secara etis. ']

# Nama kolom pendek untuk kolom pertanyaan terbuka (3 kolom terakhir)
df['Q8_Integrasi_Dosen'] = df[df.columns[13]]  # kolom ke-14
df['Q9_Risiko_AI']       = df[df.columns[14]]  # kolom ke-15
df['Q10_Kebijakan_ITSB'] = df[df.columns[15]]  # kolom ke-16

# Simpan nama kolom Likert (skala 1-4) ke dalam list
kolom_likert = ['Q1_Memahami','Q2_Tugas','Q3_Ujian','Q4_BelajarMandiri',
                'Q5_Privasi','Q6_BerpikirKritis','Q7_Kampus']

# Simpan nama kolom terbuka ke dalam list
kolom_terbuka = ['Q8_Integrasi_Dosen', 'Q9_Risiko_AI', 'Q10_Kebijakan_ITSB']

```

```python
# Hitung jumlah per jenis kelamin
print('=== Jenis Kelamin ===')
print(df['Jenis Kelamin'].value_counts())

print('\n=== Angkatan ===')
print(df['Angkatan'].value_counts())

print('\n=== Program Studi ===')
print(df['Program Studi'].value_counts())

print('\n=== Frekuensi Pakai AI ===')
print(df['Frekuensi penggunaan AI dalam 1 minggu terakhir? '].value_counts())
```

```python
# Hitung rata-rata tiap pertanyaan Likert
# Skala: 1 = Sangat Tidak Setuju, 4 = Sangat Setuju

print('=== Rata-rata Skor Likert (1-4) ===')
for kolom in kolom_likert:
    rata = df[kolom].mean()       # .mean() = rata-rata
    print(f'{kolom}: {rata:.2f}') # :.2f = tampilkan 2 angka di belakang koma

print('\n(Semakin tinggi = semakin setuju / semakin khawatir)')
```

```python
# Hitung berapa orang yang pakai tiap platform AI
# .str.contains() = cek apakah teks mengandung kata tertentu

kolom_ai = 'AI apa yang paling sering Anda gunakan untuk belajar? '
platform = ['ChatGPT', 'Gemini', 'Claude', 'Deepseek', 'Perplexity', 'Copilot']

print('=== Popularitas Platform AI ===')
for nama in platform:
    jumlah = df[kolom_ai].str.contains(nama, case=False, na=False).sum()
    persen = jumlah / len(df) * 100
    print(f'{nama}: {jumlah} orang ({persen:.0f}%)')
```

```python
# GRAFIK 4: Popularitas Platform AI (Bar Chart)

kolom_ai = 'AI apa yang paling sering Anda gunakan untuk belajar? '
platform = ['ChatGPT', 'Gemini', 'Claude', 'Deepseek', 'Perplexity', 'Copilot']

# Hitung jumlah pengguna tiap platform
jumlah_platform = []
for nama in platform:
    jumlah = df[kolom_ai].str.contains(nama, case=False, na=False).sum()
    jumlah_platform.append(jumlah)

plt.figure(figsize=(9, 5))

plt.bar(
    platform,
    jumlah_platform,
    color=['#FF6B35','#3B56D9','#A8B8FF','#D0D8FF','#E8ECF8','#F0F3FF']
)

for i, val in enumerate(jumlah_platform):
    plt.text(i, val + 0.5, str(val), ha='center', fontsize=12, fontweight='bold')

plt.title('Platform AI yang Digunakan Mahasiswa', fontsize=14, fontweight='bold')
plt.xlabel('Platform AI')
plt.ylabel('Jumlah Pengguna')
plt.ylim(0, 55)
plt.show()
```

```python
# GRAFIK 5: Rata-rata Skor Likert (Bar Horizontal)

# Label pendek untuk grafik
label_pendek = [
    'Memahami Materi',
    'Kualitas Tugas',
    'Percaya Diri Ujian',
    'Belajar Mandiri',
    'Privasi Data',
    'Berpikir Kritis',
    'Kesiapan Kampus'
]

# Hitung rata-rata tiap kolom
rata_rata = [df[kol].mean() for kol in kolom_likert]

# Urutkan dari kecil ke besar
data_sorted = sorted(zip(label_pendek, rata_rata), key=lambda x: x[1])
labels_sorted = [x[0] for x in data_sorted]
values_sorted = [x[1] for x in data_sorted]

plt.figure(figsize=(10, 6))

plt.barh(labels_sorted, values_sorted, color='#3B56D9')

# Tambahkan angka di ujung tiap batang
for i, val in enumerate(values_sorted):
    plt.text(val + 0.03, i, f'{val:.2f}', va='center', fontsize=11, fontweight='bold')

# Garis netral
plt.axvline(2.5, color='red', linestyle='--', alpha=0.5, label='Batas Netral (2.5)')

plt.title('Rata-rata Skor per Pernyataan (Skala 1-4)', fontsize=14, fontweight='bold')
plt.xlabel('Rata-rata Skor')
plt.xlim(1, 4.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# GRAFIK 6: Distribusi Jawaban (Stacked Bar)
# Grafik ini menunjukkan berapa orang jawab 1/2/3/4 tiap pertanyaan

label_pendek = [
    'Memahami\nMateri','Kualitas\nTugas','Percaya\nDiri',
    'Belajar\nMandiri','Privasi\nData','Berpikir\nKritis','Kesiapan\nKampus'
]
warna      = ['#E05C5C','#F4A261','#4CAF82','#2196F3']  # merah, oranye, hijau, biru
nama_skor  = ['Sangat Tidak Setuju (1)','Tidak Setuju (2)','Setuju (3)','Sangat Setuju (4)']

plt.figure(figsize=(13, 6))

# Stack dari bawah ke atas
posisi_bawah = [0] * len(kolom_likert)  # mulai dari 0

for skor, warna_bar, nama in zip([1, 2, 3, 4], warna, nama_skor):
    # Hitung berapa orang jawab skor ini di tiap pertanyaan
    jumlah_per_q = []
    for kol in kolom_likert:
        jumlah = (df[kol] == skor).sum()  # hitung yang sama dengan skor
        jumlah_per_q.append(jumlah)

    plt.bar(
        label_pendek,
        jumlah_per_q,
        bottom=posisi_bawah,
        color=warna_bar,
        label=nama,
        edgecolor='white',
        linewidth=1
    )

    # Tambahkan angka di dalam batang (kalau cukup besar)
    for i, (jml, bawah) in enumerate(zip(jumlah_per_q, posisi_bawah)):
        if jml >= 5:
            plt.text(i, bawah + jml/2, str(jml), ha='center', va='center',
                     fontsize=10, fontweight='bold', color='white')

    # Update posisi bawah untuk stack berikutnya
    posisi_bawah = [a + b for a, b in zip(posisi_bawah, jumlah_per_q)]

plt.title('Distribusi Jawaban Tiap Pertanyaan (n=64)', fontsize=14, fontweight='bold')
plt.ylabel('Jumlah Responden')
plt.legend(loc='upper right', fontsize=9)
plt.ylim(0, 75)
plt.tight_layout()
plt.show()
```

## **6. ANALISIS PERTANYAAN TERBUKA (3 KOLOM TERAKHIR)**

Tiga kolom terakhir berisi **jawaban bebas / pendapat** mahasiswa. Karena berbentuk teks, kita menggunakan teknik **keyword matching** — yaitu mencari kata kunci tertentu di dalam setiap jawaban untuk mengkategorikannya.

| Kolom | Pertanyaan | Tipe |
|-------|-----------|------|
| Q8 | Pendapat tentang integrasi AI oleh dosen | Pendapat |
| Q9 | Risiko terbesar AI menurut mahasiswa | Analisa |
| Q10 | Apakah ITSB perlu kebijakan AI? | Pendapat |

```python
# Lihat semua jawaban Q8
print('=== Q8: Pendapat Integrasi AI oleh Dosen ===')
for i, jawaban in enumerate(df['Q8_Integrasi_Dosen'].dropna(), 1):
    print(f'{i}. {jawaban}')
```

```python
# Kategorikan jawaban Q8 dengan keyword matching
# Ubah semua jawaban jadi huruf kecil supaya pencarian tidak case-sensitive
jawaban_q8 = df['Q8_Integrasi_Dosen'].dropna().str.lower()

# Tentukan kata kunci tiap kategori
kata_setuju = 'setuju|boleh|bisa|perlu|iya|ya|mendukung|bagus|ok'
kata_tidak  = 'tidak setuju|tidak boleh|jangan'

jumlah_setuju  = jawaban_q8.str.contains(kata_setuju).sum()
jumlah_tidak   = jawaban_q8.str.contains(kata_tidak).sum()
jumlah_syarat  = len(jawaban_q8) - jumlah_setuju - jumlah_tidak

print('=== Hasil Kategorisasi Q8 ===')
print(f' Setuju / Positif   : {jumlah_setuju} orang ({jumlah_setuju/len(jawaban_q8)*100:.0f}%)')
print(f' Bersyarat / Netral : {jumlah_syarat} orang ({jumlah_syarat/len(jawaban_q8)*100:.0f}%)')
print(f' Tidak Setuju       : {jumlah_tidak} orang ({jumlah_tidak/len(jawaban_q8)*100:.0f}%)')
```

```python
# GRAFIK 7: Pendapat Integrasi AI oleh Dosen (Donut Chart)

plt.figure(figsize=(7, 7))

plt.pie(
    [jumlah_setuju, jumlah_syarat, jumlah_tidak],
    labels=[
        f'Setuju ({jumlah_setuju} orang)',
        f'Bersyarat ({jumlah_syarat} orang)',
        f'Tidak Setuju ({jumlah_tidak} orang)'
    ],
    colors=['#4CAF82', '#F4C44A', '#E05C5C'],
    autopct='%1.0f%%',
    wedgeprops=dict(width=0.6, edgecolor='white', linewidth=2),  # width=0.6 = donut
    startangle=90,
    textprops={'fontsize': 12}
)

plt.title('Q8 — Pendapat Mahasiswa tentang\nIntegrasi AI oleh Dosen', fontsize=14, fontweight='bold')
plt.show()
```

```python
# Lihat semua jawaban Q9
print('=== Q9: Risiko Terbesar AI ===')
for i, jawaban in enumerate(df['Q9_Risiko_AI'].dropna(), 1):
    print(f'{i}. {jawaban}')
```

```python
# Kategorikan jawaban Q9 ke dalam 5 tema dengan keyword matching
jawaban_q9 = df['Q9_Risiko_AI'].dropna().str.lower()

# Definisikan tema dan kata kuncinya
tema_risiko = {
    'Berpikir Kritis Menurun' : 'kritis|bernalar|analitis|analisa',
    'Ketergantungan AI'       : 'tergantung|bergantung|ketergantungan',
    'Malas Belajar'           : 'malas|tidak belajar|tidak mau|enggan',
    'Privasi & Keamanan Data' : 'privasi|data|keamanan|bocor|kerahasiaan',
    'Plagiarisme/Kecurangan'  : 'plagiar|curang|menyontek|jujur',
}

print('=== Tema Risiko yang Disebutkan ===')
hasil_tema = {}
for nama_tema, kata_kunci in tema_risiko.items():
    jumlah = jawaban_q9.str.contains(kata_kunci, na=False).sum()
    hasil_tema[nama_tema] = jumlah
    print(f'{nama_tema}: {jumlah} orang ({jumlah/len(jawaban_q9)*100:.0f}%)')
```

```python
# GRAFIK 8: Risiko Terbesar AI (Bar Chart)

tema_nama   = list(hasil_tema.keys())
tema_jumlah = list(hasil_tema.values())

plt.figure(figsize=(10, 6))

warna_bar = ['#E05C5C','#F4A261','#F4C44A','#A8B8FF','#D0D8FF']

bars = plt.bar(tema_nama, tema_jumlah, color=warna_bar, edgecolor='white', linewidth=1.5)

# Tulis angka di atas tiap batang
for bar, val in zip(bars, tema_jumlah):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 0.3,
        f'{val} orang\n({val/64*100:.0f}%)',
        ha='center', fontsize=11, fontweight='bold'
    )

plt.title('Q9 — Risiko AI Terbesar Menurut Mahasiswa ITSB', fontsize=14, fontweight='bold')
plt.ylabel('Jumlah Responden')
plt.ylim(0, 30)
plt.xticks(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# Lihat semua jawaban Q10
print('=== Q10: Perlu Kebijakan AI di ITSB? ===')
for i, jawaban in enumerate(df['Q10_Kebijakan_ITSB'].dropna(), 1):
    print(f'{i}. {jawaban}')
```

```python
# Kategorikan jawaban Q10
jawaban_q10 = df['Q10_Kebijakan_ITSB'].dropna().str.lower()

kata_perlu  = 'perlu|ya|iya|harus|penting|setuju|diperlukan|wajib'
kata_tidak  = 'tidak perlu|tidak harus|tidak wajib|kurang perlu'

jumlah_perlu   = jawaban_q10.str.contains(kata_perlu, na=False).sum()
jumlah_tidak10 = jawaban_q10.str.contains(kata_tidak, na=False).sum()
jumlah_lain    = len(jawaban_q10) - jumlah_perlu - jumlah_tidak10

print('=== Hasil Kategorisasi Q10 ===')
print(f' Perlu Kebijakan      : {jumlah_perlu} orang ({jumlah_perlu/len(jawaban_q10)*100:.0f}%)')
print(f' Tidak Perlu          : {jumlah_tidak10} orang ({jumlah_tidak10/len(jawaban_q10)*100:.0f}%)')
print(f' Kondisional / Lainnya: {jumlah_lain} orang ({jumlah_lain/len(jawaban_q10)*100:.0f}%)')
```

```python
# GRAFIK 9: Kebijakan AI di ITSB (Donut Chart)

plt.figure(figsize=(7, 7))

plt.pie(
    [jumlah_perlu, jumlah_tidak10, jumlah_lain],
    labels=[
        f'Perlu ({jumlah_perlu} orang)',
        f'Tidak Perlu ({jumlah_tidak10} orang)',
        f'Kondisional ({jumlah_lain} orang)'
    ],
    colors=['#3B56D9', '#E05C5C', '#F4C44A'],
    autopct='%1.0f%%',
    wedgeprops=dict(width=0.6, edgecolor='white', linewidth=2),
    startangle=90,
    textprops={'fontsize': 12}
)

plt.title('Q10 — Perlu Kebijakan AI di ITSB?', fontsize=14, fontweight='bold')
plt.show()
```
